# استخراج داده
مدیر اتصال را آماده کرده است. نام ستون‌های نمونه را با ستون‌های واقعی جایگزین کنید. تاریخ‌ها حداکثر هفت روز و شامل ابتدا و پایان هستند. هر بار یک کار اجرا کنید. راهنما: docs/customer_guide_fa.docx

```python
from pyspark.sql.functions import col
from data_extractor import Query

(Query("customer_search")
 .between("2026-01-01", "2026-01-07")
 .where((col("a") > 10) & col("b").startswith("asdasa"))
 .select("a", "b", "c")
 .limit(1_000_000)
 .output_files(4)
 .run())
```

## شرط‌ها
& یعنی و، | یعنی یا و ~ یعنی نقیض. هر مقایسه را در پرانتز بنویسید؛ and و or پایتون مناسب نیستند. چند where با و ترکیب می‌شوند. تعداد فایل هدف تضمین تعداد واقعی فایل نیست. انتخاب رکوردهای limit ترتیب تضمین‌شده ندارد.

## دریافت
فرمان دریافت چاپ‌شده در پیام پایان را روی دستگاه لینوکسی آماده اجرا کنید. تاریخ فرمان تاریخ اجرای تهران است، نه تاریخ ورودی. پس از خطای مهلت، کار را مکرراً تکرار نکنید؛ قفل ممکن است تا توقف کار نگه داشته شود. برای گزارش خطا شناسه و زمان را به مدیر بدهید.
