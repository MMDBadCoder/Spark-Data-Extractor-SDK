```python
# Query را ابتدا بسازید تا تنظیم معتبر مدیر به SPARK_REMOTE اعمال شود. برای تغییر تنظیم، Kernel > Restart Kernel را انتخاب کنید.
from data_extractor import Query
query = Query("first_export")
```

```python
# یک روزی را وارد کنید که مدیر تأیید کرده در منبع موجود است. مرور ستون‌ها اختیاری است.
start_date = "2026-01-01"  # تاریخ نمونه را با یک روز موجود عوض کنید
end_date = start_date
query.between(start_date, end_date)
columns = query.show_columns()  # اختیاری؛ فقط نام و نوع ستون‌ها را نشان می‌دهد
```

```python
# اجرای اولیه بدون شرط یا انتخاب نام ستون؛ سقف را برای بررسی کوچک نگه دارید.
query.limit(10).output_files(1).run()
```
