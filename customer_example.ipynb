# پیدا کردن ستون‌ها و گرفتن خروجی
مدیر اتصال دفترچه را آماده کرده است و تنظیمات مدیر در فایل یا محیط امن بیرون دفترچه نگهداری می‌شود. کاربر لازم نیست نشانی Spark یا HDFS را وارد کند. اگر پروژه را از ZIP اجرا می‌کنید، Jupyter را از ریشهٔ پروژه اجرا کنید تا پوشهٔ `data_extractor/` کنار دفترچه باشد؛ `admin.yaml` ریشه فقط نمونهٔ خالی است و مدیر باید آن را تنظیم کند. برای محیط واقعی، پیکربندی بیرونی و محافظت‌شده ترجیح دارد. ابتدا تاریخ را وارد کنید و جدول نام و نوع ستون‌ها را ببینید؛ سپس شرط و خروجی را کامل کنید. راهنما: `docs/customer_guide_fa.docx`.

```python
from data_extractor import Query

query = Query("customer_search")
query.between("2026-01-01", "2026-01-07")
columns = query.show_columns()  # جدول فارسی نام و نوع ستون‌ها
```

## انتخاب شرط و خروجی
نام‌های `customer_id` و `status` نمونه‌اند؛ پس از دیدن جدول بالا، آن‌ها را با نام‌های واقعی عوض کنید. این سلول را پس از سلول کشف ستون‌ها اجرا کنید. `limit` و `output_files` برای خروجی لازم‌اند.

```python
from pyspark.sql.functions import col

query.where((col("customer_id") > 10) & (col("status") == "ready"))
query.select("customer_id", "status")
query.limit(100_000)
query.output_files(4)
query.run()
```

## راهنمای شرط‌ها
`&` یعنی و، `|` یعنی یا، و `~` یعنی نقیض. هر مقایسه را در پرانتز بنویسید؛ چند `query.where(...)` با و ترکیب می‌شوند. تعداد فایل واقعی ممکن است با تعداد هدف فرق کند. انتخاب ردیف‌های `limit` ترتیب تضمین‌شده ندارد.
